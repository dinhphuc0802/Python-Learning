```python
email_go = input("Nhập email: ")
```

```python
email_sach = email_go.strip().lower()
```

```python
hop_le = "@" in email_sach
print(f"Hợp lệ: {hop_le}")
```

```python
if hop_le:
  username, domain = email_sach.split("@")
  print(f"Email sạch: {email_sach}")
  print(f"username: {username}")
  print(f"domain: {domain}")
else:
  print("Email không hợp lệ")
```

```python
thue_tuan = float(input("Nhập giá thuê 1 tuần: "))
```

```python
income = float(input("Nhập lương 1 thang: "))
```

```python
thue_thang = thue_tuan * 4.33
```

```python
phan_tram = thue_thang/income
```

```python
thue_nam = thue_thang * 12

print(f"Thuê/tháng: ${thue_thang:,.2f}")
print(f"Thuê/năm: ${thue_nam:,.2f}")
print(f"Phần trăm: {phan_tram:.1%}")
kn = "Quá cao!" if phan_tram >= 0.3 else "Hợp lí"
print(f"Khuyến nghị: {kn}")
```

```python
doanh_thu = float(input("Nhập doanh thu: "))
chi_phi = float(input("Nhập chi phí: "))
luy_ke = doanh_thu - chi_phi
roi = (luy_ke/chi_phi)*100
```

```python
print(f"Lợi nhuận: ${luy_ke:,.2f}")
print(f"ROI: {roi:.1f}%")
if roi > 100:
  print("Lợi nhuận xuất sắc")
elif roi >= 50:
  print("Lợi nhuận tốt")
elif roi >= 0:
  print("Lợi nhuận trung bình")
else:
  print("Thua lỗ")
```
